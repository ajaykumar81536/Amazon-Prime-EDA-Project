## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Create a fresh copy before performing data wrangling
titles_df = titles_raw.copy()
credits_df = credits_raw.copy()

# Remove duplicate rows from both datasets
titles_df = titles_df.drop_duplicates()
credits_df = credits_df.drop_duplicates()

# Remove square brackets and quotes from genres
titles_df['genres'] = titles_df['genres'].str.replace('[', '', regex=False)
titles_df['genres'] = titles_df['genres'].str.replace(']', '', regex=False)
titles_df['genres'] = titles_df['genres'].str.replace("'", '', regex=False)

# Remove square brackets and quotes from production countries
titles_df['production_countries'] = titles_df['production_countries'].str.replace('[', '', regex=False)
titles_df['production_countries'] = titles_df['production_countries'].str.replace(']', '', regex=False)
titles_df['production_countries'] = titles_df['production_countries'].str.replace("'", '', regex=False)

# Fill missing age certifications
titles_df['age_certification'] = titles_df['age_certification'].fillna('Not Rated')

# Seasons are not applicable to movies, so replace missing values with 0
titles_df['seasons'] = titles_df['seasons'].fillna(0)

# Display the first five rows after wrangling
titles_df.head()
````

### What all manipulations have you done and insights you found?

The following data-wrangling steps were performed to prepare the datasets for analysis:

- Duplicate records were removed from both datasets to avoid repeated records during analysis.

- Square brackets and quotation marks were removed from the `genres` and `production_countries` columns to make their values easier to work with.

- Missing values in `age_certification` were replaced with `Not Rated` so that missing certification records can be treated as a separate category.

- Missing values in the `seasons` column were replaced with `0` because seasons are not applicable to movies.

After these basic cleaning steps, the datasets are more suitable for further analysis and visualization.
